```python
import streamlit as st
import pandas as pd
import plotly.express as px

# Configuração da página
st.set_page_config(
    page_title="Análise de Acesso à Internet no Brasil",
    layout="wide"
)

# Título e Apresentação (Aula 4)
st.title("📊 Dashboard: Motivos do Não Acesso à Internet no Brasil")
st.write("""
Este dashboard apresenta uma análise baseada em dados públicos sobre o acesso à Internet no Brasil. 
O objetivo é compreender as barreiras digitais enfrentadas pela população e as diferenças entre as regiões do país.
""")

# Upload do arquivo CSV (Aula 2)
arquivo = st.file_uploader("Envia um ficheiro CSV com os dados de acesso", type=["csv"])

if arquivo is not None:
    # Leitura dos dados com Pandas
    df = pd.read_csv(arquivo)

    # Tratamento básico dos dados (Aula 2)
    df = df.drop_duplicates()
    if 'Motivo_Nao_Acesso' in df.columns:
        df['Motivo_Nao_Acesso'] = df['Motivo_Nao_Acesso'].astype(str).str.strip()

    # Informações Gerais e Métricas/Indicadores (Aula 2 e Aula 3)
    st.subheader("📌 Visão Geral dos Dados")
    col1, col2, col3 = st.columns(3)
    col1.metric("Total de Registos", df.shape[0])
    col2.metric("Quantidade de Colunas", df.shape[1])
    
    if "Percentual_Pessoas" in df.columns:
        media_perc = df["Percentual_Pessoas"].mean()
        col3.metric("Média Percentual Apontada", f"{round(media_perc, 2)}%")

    st.markdown("---")

    # Visualização dos dados brutos
    with st.expander("Ver primeiras linhas da base de dados"):
        st.dataframe(df.head())

    # --- GRÁFICO 1: Motivos do Não Acesso (Aula 3) ---
    st.subheader("1. Principal Motivo do Não Acesso por Região")
    if 'Motivo_Nao_Acesso' in df.columns and 'Percentual_Pessoas' in df.columns:
        fig1 = px.bar(
            df, 
            x="Motivo_Nao_Acesso", 
            y="Percentual_Pessoas", 
            color="Regiao",
            barmode="group",
            title="Distribuição dos Motivos de Não Acesso à Internet (%)",
            labels={"Motivo_Nao_Acesso": "Motivo", "Percentual_Pessoas": "Percentual (%)"}
        )
        st.plotly_chart(fig1, use_container_width=True)

        # Explicação do Gráfico 1 (Aula 4)
        st.markdown("""
        > **O que podemos observar?**  
        > Os dados indicam que, nas regiões mais desenvolvidas e urbanizadas (como o Sul e o Sudeste), o principal motivo alegado para não utilizar a Internet é a **falta de conhecimento/não saber usar**.  
        > Por outro lado, nas regiões Norte e Nordeste, o **custo do serviço** e a **indisponibilidade da infraestrutura de rede** ganham um peso significativamente maior.
        """)

    st.markdown("---")

    # --- GRÁFICO 2: Distribuição por Região (Aula 3) ---
    st.subheader("2. Comparação Média do Percentual do Problema por Região")
    if 'Regiao' in df.columns and 'Percentual_Pessoas' in df.columns:
        df_regiao = df.groupby('Regiao')['Percentual_Pessoas'].mean().reset_index()
        fig2 = px.bar(
            df_regiao, 
            x="Regiao", 
            y="Percentual_Pessoas", 
            color="Regiao",
            title="Média dos Percentuais Registados por Região (%)",
            labels={"Regiao": "Região", "Percentual_Pessoas": "Média (%)"}
        )
        st.plotly_chart(fig2, use_container_width=True)

        # Explicação do Gráfico 2 (Aula 4)
        st.markdown("""
        > **O que podemos observar?**  
        > A comparação regional evidencia que as regiões Norte e Nordeste concentram médias percentuais mais altas associadas às barreiras de acesso. Isso aponta para uma desigualdade socioespacial no acesso a bens e serviços digitais no país.
        """)

    st.markdown("---")

    # --- EXEMPLO DE USO DA INTELIGÊNCIA ARTIFICIAL (Aula 5) ---
    st.subheader("🤖 Exemplo de Uso da Inteligência Artificial no Projeto")
    st.markdown("""
    * **Problema encontrado:** Necessidade de agrupar e filtrar os dados por região no Plotly de forma responsiva sem gerar erros de tipo[cite: 18, 19].
    * **Prompt utilizado:** *"Como criar um gráfico de barras agrupado por região usando Plotly Express e Streamlit garantindo que valores numéricos sejam tratados corretamente?"*[cite: 12, 19]
    * **Sugestão da IA:** A IA sugeriu a conversão explícita com `pd.to_numeric` e o uso do parâmetro `barmode='group'` no Plotly Express[cite: 10, 12, 19].
    * **Validação do Grupo:** Testou-se a aplicação com o upload de ficheiros no formato CSV e verificou-se se os valores agregados correspondiam exatamente aos totais calculados no Pandas[cite: 17, 19, 23].
    """)

else:
    st.info("Por favor, envia um ficheiro `.csv` acima para visualizar a análise.")
```
